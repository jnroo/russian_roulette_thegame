```python
import time as time
import random as random
 
print("oyuna hoşgeldin. bugüne kadar yaptığın tüm seçimler seni buraya getirdi.")

class Rulet():
    can1 = 3
    can2 = 3
    name1 = ""
    name2 = ""
    turn = 0
    shots_left_for_death = 0

 
    def setup(name1, name2):
        Rulet.can1 = 3
        Rulet.can2 = 3
        Rulet.name1 = name1
        Rulet.name2 = name2 
        Rulet.yeni_el()

    def yeni_el():
        Rulet.turn = 0
        Rulet.shots_left_for_death = random.randint(0, 5)
        time.sleep(2)
        first_shot_trigger = input("\nhazır olduğun zaman hazırım veya hazır yaz. tereddüt edenler hoş karşılanmaz: ")
        
        if first_shot_trigger.lower() in ["hazırım", "hazır"]:
            Rulet.first_shot()
        else:
            print("üzgünüm.. ya da değilim. \ngörüşmemek üzere.")
            Rulet.yeni_el()


    def first_shot():
        time.sleep(3)
        print("buraya kadar gelmiş olman beni şaşırttı.")
        time.sleep(3)
        print("tetiği çekiyorsun. buradan geri dönüş yok.")
        Rulet.shots_left_for_death -= 1
        
        if Rulet.shots_left_for_death < 0:
            print("yazık oldu, kirayı ödeyememenin bedeli bu dünyada bu olmamalıydı, dışarı çıkarın.")
            Rulet.can1 -= 1
            print(Rulet.name1 + "in " + str(Rulet.can1) + " ve " + Rulet.name2 + "nin " + str(Rulet.can2) + " canı kaldı.")
            if Rulet.can1 == 0:
                print(Rulet.name1 + " oyunu kaybetti")
                return
            Rulet.yeni_el()
        else:
            print("kovan boştu. sıra " + Rulet.name2 + "'ye geçti")
            time.sleep(3)
            Rulet.turn += 1
            Rulet.next_shot()

            
    def next_shot():
        time.sleep(2)
        if Rulet.turn % 2 == 1:
            current_player = Rulet.name2
        else:
            current_player = Rulet.name1
            
        print("sıra " + current_player + " de ve tetiği çekiyor.")
        time.sleep(2)
        Rulet.shots_left_for_death -= 1
        
        if Rulet.shots_left_for_death < 0:
            print(current_player + "in son günüymüş. dışarı çıkarın ve sıradaki gelsin.")
            if current_player == Rulet.name1:
                Rulet.can1 -= 1
            else:
                Rulet.can2 -= 1
            print(Rulet.name1 + "'nin " + str(Rulet.can1) + " ve " + Rulet.name2 + "nin " + str(Rulet.can2) + " canı kaldı.")
            
            if Rulet.can1 == 0:
                print(Rulet.name1 + " oyunu kaybetti")
                return
            elif Rulet.can2 == 0:
                print(Rulet.name2 + " oyunu kaybetti")
                return
                
            Rulet.yeni_el() 
        else:
            print("boş kovan. sıra diğer kişiye geçiyor")
            Rulet.turn += 1
            Rulet.next_shot()

oyuncu_ismi1 = input("1. ismi gir: ")
oyuncu_ismi2 = input("2. ismi gir: ")
Rulet.setup(oyuncu_ismi1,oyuncu_ismi2)
```
